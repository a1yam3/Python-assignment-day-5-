```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("Titanic Dataset.xlsx")
```

```python
#survived distribuition
print(df['Survived'].value_counts())
```

```python
# Survival rate by gender
gender_survival = pd.crosstab(df['Gender'], df['Survived'], normalize='index') * 100
print("\nSurvival Rate by Gender (%):")
print(gender_survival)
```

```python
# Survival rate by passenger class
class_survival = pd.crosstab(df['Pclass'], df['Survived'], normalize='index') * 100
print("\nSurvival Rate by Passenger Class (%):")
print(class_survival)
```

```python
# countplot of Survival Distribution 
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Survived')
plt.title('Survival Distribution')
plt.tight_layout()
plt.show()
```

```python
# countplot of Survival by Gender
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Gender', hue='Survived')
plt.title('Survival by Gender')
plt.tight_layout()
plt.show()
```

```python
# countplot of Survival by Passenger Class
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Survival by Passenger Class')
plt.tight_layout()
plt.show()
```

```python
print("\nCONCLUSION 1:")
print("Female passengers had a significantly higher survival rate than male passengers.")
```

```python
print("\nCONCLUSION 2:")
print("First-class passengers had a much higher survival rate than third-class passengers.")
```
